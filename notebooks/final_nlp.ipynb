```python
y_train=pd.DataFrame(y_train,columns=['labels'])
y_emotion = pd.DataFrame(y_emotion, columns=['labels'])
y_hate = pd.DataFrame(y_hate, columns=['labels'])
y_irony = pd.DataFrame(y_irony, columns=['labels'])
y_offensive = pd.DataFrame(y_offensive, columns=['labels'])
y_sentiment = pd.DataFrame(y_sentiment, columns=['labels'])
```

```python
import re
```

```python
def clean_text(text):

    text = re.sub(r'[^A-Za-z0-9\s]', '', text)
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"#", "", text)
    return text.lower()

x_train['text'] = x_train['text'].apply(clean_text)
```

```python
from sentence_transformers import  SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
embeddings = model.encode(x_train['text'].tolist(), show_progress_bar=True)
```

```python
model.save('sentence_transformer_model.h5')
```

```python
m=0
for i in embeddings:
  m=max(m,len(i))
```

```python
from sklearn.model_selection import train_test_split

# Use the processed embeddings directly
x_train, x_test, y_train, y_test = train_test_split(embeddings, y_train, test_size=0.2, random_state=42,shuffle=True)
```

# LSTM

```python
from tensorflow.early_stopping import Early_Stopping
```

```python
## train lstm
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Embedding,LSTM,Dropout,GRU

model1=Sequential()
model1.add(LSTM(64, input_shape=(1, 384)))
model1.add(Dense(20,activation='relu'))
model1.add(Dropout(0.2))
model1.add(Dense(33,activation='softmax'))

model1.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model1.summary()
```

```python
from tensorflow.keras.utils import to_categorical

y_train_encoded = to_categorical(y_train, num_classes=33)
y_test_encoded = to_categorical(y_test, num_classes=33)
```

```python
x_train_reshaped = x_train.reshape(x_train.shape[0], 1, x_train.shape[1])
x_test_reshaped = x_test.reshape(x_test.shape[0], 1, x_test.shape[1])
```

```python


history=model1.fit(x_train_reshaped,y_train_encoded,epochs=15,validation_data=(x_test_reshaped,y_test_encoded),verbose=1)
```

```python
def prdict_next_word(model,tokenizer,text):
    token_list=tokenizer.encode([text])
    token_list=token_list.reshape(token_list.shape[0],1,token_list.shape[1])
    predicted=model.predict(token_list,verbose=0)
    predicted_word_index=np.argmax(predicted,axis=1)  ## This line means: which word have high probability take that index

    return predicted_word_index
```

```python
model1.evaluate(x_test_reshaped, y_test_encoded)
```

```python
inp='people walking with crutches '
print("op: ",prdict_next_word(model1,model,inp))
```

# bidirectional lstm

```python
from keras.models import Sequential
from keras.layers import Dropout, Dense, LSTM, Bidirectional,Embedding, GlobalMaxPool1D
embed_len=128


model2 = Sequential()
model2.add(Bidirectional(LSTM(80,return_sequences=True)))
model2.add(Bidirectional(LSTM(80,return_sequences=True)))
model2.add(GlobalMaxPool1D())
model2.add(Dropout(0.5))
model2.add(Dense(64, activation="relu"))
model2.add(Dropout(0.5))
model2.add(Dense(33, activation="softmax"))

model2.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(monitor="val_loss", mode="min", patience=0,verbose=1)

model2.compile(loss = 'categorical_crossentropy',
             optimizer = "adam",
             metrics=["accuracy"])
```

```python
batch_size=64
epochs = 15

model2.fit(x_train_reshaped, y_train_encoded, batch_size=batch_size, epochs = epochs,
          shuffle=True,validation_data=(x_test_reshaped, y_test_encoded), callbacks=[early_stop])
```

```python
model2.save('bidirectionalRNN.h5')
```

```python
model2.evaluate(x_test_reshaped, y_test_encoded)
```
